O Jupyter utilizado é o do contêiner `spark` (imagem construída a partir de `spark/Dockerfile`, orquestrada por `spark/docker-compose.yml`), publicado no host na porta `8889`.

## a) Inicialização e acesso ao Jupyter (token ocultado)

Inicialização do contêiner e localização do token nos logs, no terminal do host (pasta `spark/`). O `sed` substitui o valor do token antes de exibi-lo:

```bash
docker compose up -d
docker compose logs spark | grep -m2 -E 'Jupyter Server .* is running|token=' | sed -E 's/token=[a-f0-9]+/token=****/'
```

Saída:

```
spark-1  | [I 2026-11-04 10:25:40.422 ServerApp] Jupyter Server 2.8.0 is running at:
spark-1  | [I 2026-11-04 10:25:40.422 ServerApp] http://7055018ba985:8888/lab?token=****
```

O acesso foi feito em `http://localhost:8889` com esse token. As células abaixo executam **dentro** do servidor Jupyter.

````python
# Servidor Jupyter em execução (token mascarado)
!jupyter server list | sed -E 's/token=[a-f0-9]+/token=****/'
````
